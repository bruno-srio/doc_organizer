### Organizando documentos históricos com algoritmos

```python
# Número de documentos
N = int(input("Introduza o número de documentos: "))
# Número de relações conhecidas
L = int(input("Introduza o número de relações conhecidas: "))

# Criando uma lista de listas para representar o grafo
# Cada elemento da lista representa um node, e a lista dentro dele as arestas
grafo = [[] for i in range(N)]

# Adicionando as arestas ao grafo, baseado nas relações conhecidas
for i in range(L):
    u, v = map(int, input("Separando os números por um espaço, introduza os pares conhecidos: ").split())
    # Decrementando em 1 para ajustar a indexação do array (0 a N-1)
    u -= 1
    v -= 1
    grafo[u].append(v)

# Criando uma lista para armazenar a ordem temporal dos documentos
ordem = []

# Criando uma lista para armazenar os nodes visitados
visitado = [False for i in range(N)]

# Criando uma lista para armazenar o número de arestas de entrada de cada node
num_entradas = [0 for i in range(N)]

# Contando o número de arestas de entrada de cada node
for u in range(N):
    for v in grafo[u]:
        num_entradas[v] += 1

# Implementando o algoritmo DFS
def dfs(node):
    # Marcar node atual como visitado
    visitado[node] = True
    # Para cada aresta (v) do node atual (u)
    for v in grafo[node]:
        # Se o node v ainda não foi visitado
        if not visitado[v]:
            # Visitar o node v
            dfs(v)
    # Adiciona o node atual à ordem temporal
    ordem.append(node)

# Aplicando o DFS em todos os nodes
for i in range(N):
    if not visitado[i]:
        dfs(i)

# Verificando se é possível determinar uma ordem temporal coerente
if len(ordem) != N:
    print("Incoerente")
# Verificando se a informação é insuficiente
elif any(x == 0 for x in num_entradas):
    print("Insuficiente")  

else:
    # Incrementando em 1 para ajustar a indexação (1 a N)
    ordem = [x + 1 for x in ordem[::-1]]
    print(*ordem)
```
